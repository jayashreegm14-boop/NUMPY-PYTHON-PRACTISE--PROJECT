```python
# add bonus mark to SQL but cap the result at 100
import numpy as np

students = np.array(["Arun", "Priya", "Kumar", "Divya", "Rahul"])

# Marks: Python, SQL, Maths
marks = np.array([
    [85, 90, 80],
    [75, 65, 85],
    [95, 98, 90],
    [60, 70, 65],
    [80, 95, 75]
])

# Add 5 bonus marks to SQL (second column)
marks[:, 1] = np.minimum(marks[:, 1] + 5, 100)

print("Updated Marks:")
print(marks)
```

```python
# replace any missing values in anew copy of the dataset with that subject mean
import numpy as np

students = np.array(["Arun", "Priya", "Kumar", "Divya", "Rahul"])

# Marks: Python, SQL, Maths
marks = np.array([
    [85, 90, 80],
    [75, np.nan, 85],
    [95, 80, 90],
    [60, 70, 65],
    [80, 95, np.nan]
])

# Create a new copy of dataset
new_marks = marks.copy()

# Replace missing values using square brackets []
new_marks[np.isnan(new_marks)] = 100
print("Updated Marks:")
print(new_marks)
```

```python
#sort student by average score,highest first
import numpy as np

students = np.array(["Arun", "Priya", "Kumar", "Divya", "Rahul"])

# Marks: Python, SQL, Maths
marks = np.array([
    [85, 90, 80],
    [75, 65, 85],
    [95, 80, 90],
    [60, 70, 65],
    [80, 95, 75]
])

# Calculate average for each student
average = np.mean(marks, axis=1)

# Sort average in descending order
index = np.argsort(average)[::-1]

# Display sorted students
sorted_students = np.max(index)

print("Students sorted by highest average:",students[index])
print(sorted_students)
```

```python
#calculate the percentage of the students whose average is at least 60
import numpy as np
students = np.array(["Arun", "Priya", "Kumar", "Divya", "Rahul"])

# Marks: Python, SQL, Maths
marks = np.array([
    [85, 90, 80],
    [75, 65, 85],
    [95, 80, 90],
    [60, 70, 65],
    [80, 95, 75]
])

# Calculate student averages
average = np.mean(marks, axis=1)

# Check average >= 60
result = average >= 60

# Calculate percentage
percentage = (np.sum(result) / len(students)) * 100

print("Students with average >= 60:", np.sum(result))
print("Total Students:", len(students))
print("Percentage:", percentage, "%")
```
